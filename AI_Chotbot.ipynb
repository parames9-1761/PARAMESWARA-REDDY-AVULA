```python
# ============================================================
# 🤖 AI CHATBOT USING PYTHON - COMPLETE PROJECT IN ONE CELL
# NLP + MACHINE LEARNING + CHAT INTERFACE
# ============================================================

# Install required libraries
!pip -q install nltk scikit-learn

# ------------------------------------------------------------
# 1. IMPORT LIBRARIES
# ------------------------------------------------------------

import re
import random
import nltk
import datetime
import warnings

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

warnings.filterwarnings("ignore")

# Download NLTK data
nltk.download("punkt", quiet=True)

# ------------------------------------------------------------
# 2. TRAINING DATA
# ------------------------------------------------------------

training_data = {

    "greeting": [
        "hello",
        "hi",
        "hey",
        "hello chatbot",
        "hi chatbot",
        "hey there",
        "good morning",
        "good afternoon",
        "good evening",
        "how are you",
        "how are you doing",
        "nice to meet you"
    ],

    "goodbye": [
        "bye",
        "goodbye",
        "see you",
        "see you later",
        "talk to you later",
        "exit",
        "quit",
        "good night"
    ],

    "thanks": [
        "thank you",
        "thanks",
        "thank you so much",
        "thanks a lot",
        "that was helpful",
        "thank you chatbot"
    ],

    "name": [
        "what is your name",
        "who are you",
        "tell me your name",
        "what should I call you",
        "who am I talking to",
        "introduce yourself"
    ],

    "creator": [
        "who created you",
        "who made you",
        "who developed you",
        "who programmed you",
        "who is your creator"
    ],

    "python": [
        "what is python",
        "tell me about python",
        "what is python programming",
        "why learn python",
        "python programming language",
        "what can python do"
    ],

    "ai": [
        "what is artificial intelligence",
        "what is ai",
        "tell me about ai",
        "explain artificial intelligence",
        "what does ai mean"
    ],

    "machine_learning": [
        "what is machine learning",
        "what is ml",
        "explain machine learning",
        "tell me about machine learning",
        "how does machine learning work"
    ],

    "nlp": [
        "what is nlp",
        "what is natural language processing",
        "explain nlp",
        "tell me about nlp",
        "what does nlp mean"
    ],

    "chatbot": [
        "what is a chatbot",
        "what is chatbot",
        "how does chatbot work",
        "tell me about chatbots",
        "explain chatbot"
    ],

    "project": [
        "tell me about this project",
        "what is this project",
        "explain your project",
        "what technologies are used",
        "how was you created"
    ],

    "study": [
        "help me study",
        "i want to study",
        "can you help with studies",
        "study help",
        "help me learn",
        "i need help with learning"
    ],

    "joke": [
        "tell me a joke",
        "make me laugh",
        "say something funny",
        "tell a funny joke",
        "give me a joke"
    ],

    "time": [
        "what time is it",
        "tell me the time",
        "current time",
        "what is the time now",
        "time now"
    ],

    "date": [
        "what is today's date",
        "tell me today's date",
        "what date is it",
        "current date",
        "today's date"
    ],

    "help": [
        "help",
        "what can you do",
        "what are your features",
        "how can you help me",
        "what can you help with"
    ]
}

# ------------------------------------------------------------
# 3. RESPONSE DATA
# ------------------------------------------------------------

responses = {

    "greeting": [
        "Hello! 👋 How can I help you?",
        "Hi! 😊 Nice to meet you.",
        "Hey! 🤖 What can I do for you?",
        "Hello! I'm ready to chat with you."
    ],

    "goodbye": [
        "Goodbye! 👋 Have a great day!",
        "See you later! 😊",
        "Bye! Take care!",
        "It was nice talking to you!"
    ],

    "thanks": [
        "You're welcome! 😊",
        "No problem! 👍",
        "Happy to help! 🤖",
        "Anytime!"
    ],

    "name": [
        "My name is PyBot 🤖.",
        "I'm PyBot, an AI chatbot built using Python.",
        "You can call me PyBot!"
    ],

    "creator": [
        "I was created as a Python AI chatbot project.",
        "I'm built using Python, NLP and machine learning.",
        "My creator is the developer who built this project using Python."
    ],

    "python": [
        "Python is a high-level programming language known for its simple syntax and huge collection of libraries.",
        "Python is widely used in AI, machine learning, web development, automation and data science.",
        "Python is one of the most popular programming languages for AI and data science."
    ],

    "ai": [
        "Artificial Intelligence, or AI, is technology that allows computers to perform tasks that normally require human intelligence.",
        "AI can be used for vision, speech recognition, decision-making, language processing and more.",
        "AI is a field of computer science focused on creating intelligent systems."
    ],

    "machine_learning": [
        "Machine Learning is a branch of AI where computers learn patterns from data and make predictions or decisions.",
        "ML allows a computer to learn from examples instead of being explicitly programmed for every situation.",
        "Common machine learning methods include classification, regression and clustering."
    ],

    "nlp": [
        "NLP stands for Natural Language Processing. It helps computers understand and process human language.",
        "NLP is used in chatbots, translation, sentiment analysis and voice assistants.",
        "NLP allows computers to work with human language."
    ],

    "chatbot": [
        "A chatbot is a software program that communicates with users through text or voice.",
        "Chatbots can use rules, machine learning or large language models to generate responses.",
        "I'm an example of a machine-learning based Python chatbot."
    ],

    "project": [
        "This is an AI chatbot project developed using Python, NLP, TF-IDF and machine learning.",
        "The chatbot classifies user messages into intents and selects an appropriate response.",
        "The main technologies are Python, NLP, TF-IDF Vectorization and Logistic Regression."
    ],

    "study": [
        "Sure! 📚 Tell me what subject or topic you want to study.",
        "Of course! I can help you understand programming, AI, DBMS, operating systems and many other topics.",
        "Let's study! Tell me your topic."
    ],

    "joke": [
        "Why do programmers prefer dark mode? Because light attracts bugs! 😂",
        "Why did the computer go to the doctor? Because it had a virus! 🤣",
        "A SQL query walks into a bar and asks: Can I JOIN you? 😄"
    ],

    "help": [
        "I can answer questions about Python, AI, ML, NLP and chatbots. I can also tell you the date and time.",
        "Try asking: 'What is Python?', 'What is AI?', 'What is NLP?', or 'Tell me a joke'.",
        "I'm here to help with basic AI and programming questions."
    ]
}

# ------------------------------------------------------------
# 4. CREATE TRAINING DATA
# ------------------------------------------------------------

sentences = []
labels = []

for intent, examples in training_data.items():
    for sentence in examples:
        sentences.append(sentence)
        labels.append(intent)

# ------------------------------------------------------------
# 5. CREATE MACHINE LEARNING MODEL
# ------------------------------------------------------------

model = Pipeline([
    ("tfidf", TfidfVectorizer(
        lowercase=True,
        ngram_range=(1, 2),
        sublinear_tf=True
    )),
    ("classifier", LogisticRegression(
        max_iter=2000
    ))
])

# Train the chatbot
model.fit(sentences, labels)

print("✅ Chatbot model trained successfully!")
print("📊 Training examples:", len(sentences))
print("🧠 Intents:", len(set(labels)))

# ------------------------------------------------------------
# 6. TEXT PREPROCESSING
# ------------------------------------------------------------

def clean_text(text):

    text = text.lower()

    # Remove unnecessary spaces
    text = re.sub(r"\s+", " ", text)

    # Remove special characters
    text = re.sub(r"[^a-zA-Z0-9\s]", "", text)

    return text.strip()


# ------------------------------------------------------------
# 7. GET INTENT
# ------------------------------------------------------------

def get_intent(user_input):

    cleaned = clean_text(user_input)

    probabilities = model.predict_proba([cleaned])[0]

    best_index = probabilities.argmax()

    confidence = probabilities[best_index]

    intent = model.classes_[best_index]

    return intent, confidence


# ------------------------------------------------------------
# 8. CHATBOT RESPONSE FUNCTION
# ------------------------------------------------------------

def chatbot_response(user_input):

    if not user_input.strip():
        return "Please type something so I can respond. 😊"

    intent, confidence = get_intent(user_input)

    # Confidence threshold
    if confidence < 0.35:

        return (
            "I'm not completely sure what you mean. 🤔\n"
            "Try asking me about Python, AI, ML, NLP or chatbots."
        )

    # Time
    if intent == "time":

        current_time = datetime.datetime.now().strftime("%I:%M:%S %p")

        return f"The current time is {current_time}. ⏰"

    # Date
    if intent == "date":

        current_date = datetime.datetime.now().strftime("%d-%m-%Y")

        return f"Today's date is {current_date}. 📅"

    # Normal responses
    return random.choice(responses[intent])


# ------------------------------------------------------------
# 9. CHAT INTERFACE
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("🤖 PYBOT - AI CHATBOT USING PYTHON")
print("=" * 60)

print("\nChatbot is ready! 🎉")

print("\nYou can ask things like:")
print("👉 Hello")
print("👉 What is Python?")
print("👉 What is AI?")
print("👉 What is Machine Learning?")
print("👉 What is NLP?")
print("👉 Tell me a joke")
print("👉 What time is it?")
print("👉 What is today's date?")
print("👉 Help")

print("\nType 'bye', 'exit' or 'quit' to stop.")
print("=" * 60)

# ------------------------------------------------------------
# 10. START CHAT
# ------------------------------------------------------------

while True:

    user_input = input("\n👤 You: ")

    # Exit commands
    if user_input.lower().strip() in [
        "bye",
        "goodbye",
        "exit",
        "quit"
    ]:

        print("🤖 PyBot:", random.choice(responses["goodbye"]))
        break

    # Generate response
    answer = chatbot_response(user_input)

    print("🤖 PyBot:", answer)

# ============================================================
# END OF PROJECT
# ============================================================
```
